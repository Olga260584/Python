**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

### 1. Формулювання гіпотез

**Нульова гіпотеза H₀:** μ = 123,7 хв - середній час до покупки не збільшився порівняно з історичним значенням.

**Альтернативна гіпотеза Hₐ:** μ > 123,7 хв - середній час до покупки після редизайну збільшився.

**Тип тесту:** правосторонній одновибірковий t-тест Стьюдента, оскільки вибірка мала (n = 7), а стандартне відхилення генеральної сукупності невідоме.

```python

import numpy as np

data = np.array([128, 135, 121, 142, 126, 151, 123])

sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)
sample_size = len(data)

print("Середнє значення:", sample_mean)
print("Вибіркове стандартне відхилення:", sample_std)
print("Розмір вибірки:", sample_size)
```

```python

from scipy import stats

population_mean = 123.7
alpha = 0.10

t_stat, p_value = stats.ttest_1samp(
    data,
    popmean=population_mean,
    alternative='greater'
)

df = sample_size - 1
t_critical = stats.t.ppf(1 - alpha, df)

print("t-статистика:", t_stat)
print("Критичне t-значення:", t_critical)
print("p-value:", p_value)
```

### 4. Прийняття рішення за p-value

Отримане p-value = 0,042.

Рівень значущості α = 0,10.

Оскільки p-value < α (0,042 < 0,10), ми відхиляємо нульову гіпотезу H₀.

Отже, маємо достатньо статистичних доказів на рівні значущості 10%, щоб стверджувати, що середній час до покупки після редизайну збільшився.

### 5. Висновок

До редизайну середній час до покупки становив 123,7 хвилини, а після редизайну у вибірці з 7 покупців - 132,29 хвилини.

Середній час збільшився приблизно на 8,59 хвилини.

За результатами одновибіркового правостороннього t-тесту (p-value = 0,042 < α = 0,10) виявлено статистично значуще збільшення середнього часу до покупки.

Отже, результати свідчать про збільшення часу до покупки після редизайну сайту. Проте сам по собі цей тест не доводить, що причиною збільшення став саме редизайн.

### 6. Перевірка при рівні значущості α = 0,05

При зміні рівня значущості з α = 0,10 на α = 0,05 значення p-value залишається незмінним і становить приблизно 0,042.

Оскільки p-value = 0,042 < 0,05, ми також відхиляємо нульову гіпотезу H₀.

**Висновок:** рішення не змінюється. Збільшення середнього часу до покупки є статистично значущим як на рівні значущості 10%, так і на рівні 5%.

### Завдання 2. Перевірка гіпотези для пропорції

**Нульова гіпотеза H₀:** p = 0,23 - частка дорослих, які впізнають логотип, не збільшилася порівняно з попереднім рівнем.

**Альтернативна гіпотеза Hₐ:** p > 0,23 - частка дорослих, які впізнають логотип, збільшилася.

**Тип тесту:** правосторонній одновибірковий Z-тест для пропорції.

**Рівень значущості:** α = 0,03.

```python

# Дані задачі
n = 1200
successes = 311
p0 = 0.23
alpha = 0.03

# Частка людей, які впізнали логотип
p_hat = successes / n

print("Частка після кампанії:", p_hat)
print("Відсоток після кампанії:", p_hat * 100)
```

```python

from statsmodels.stats.proportion import proportions_ztest

z_stat, p_value = proportions_ztest(
    count=successes,
    nobs=n,
    value=p0,
    alternative='larger'
)

print("Z-статистика:", z_stat)
print("p-value:", p_value)
```

### Висновок за результатами Z-тесту

За результатами Z-тесту для однієї пропорції отримано:

- Z-статистика = 2,306
- p-value = 0,01056
- Рівень значущості α = 0,03

Оскільки p-value < α (0,01056 < 0,03), нульову гіпотезу H₀ відхиляємо.

Отже, на рівні значущості 3% маємо достатньо статистичних доказів того, що частка дорослих, які впізнають логотип компанії, перевищує попередній рівень 23%.

У вибірці після рекламної кампанії впізнаваність становила приблизно 25,92%.

```python

from statsmodels.stats.proportion import proportion_confint

ci_low, ci_high = proportion_confint(
    count=successes,
    nobs=n,
    alpha=0.03,
    method='wilson'
)

print("Нижня межа:", ci_low)
print("Верхня межа:", ci_high)
print("Довірчий інтервал у %:",
      ci_low * 100, "-", ci_high * 100)
```

### Довірчий інтервал та інтерпретація

За допомогою бібліотеки `statsmodels` методом Вілсона обчислено двосторонній 97% довірчий інтервал для частки дорослих, які впізнають логотип компанії.

**Отриманий довірчий інтервал: [23,27%; 28,75%].**

Цей інтервал показує діапазон правдоподібних значень справжньої частки дорослих, які впізнають логотип після рекламної кампанії.

Попереднє значення 23% знаходиться нижче нижньої межі довірчого інтервалу (23,27%).

Це додатково підтверджує результат Z-тесту, за яким p-value = 0,01056 < α = 0,03.

**Загальний висновок:** на рівні значущості 3% маємо достатньо статистичних доказів того, що впізнаваність логотипу після рекламної кампанії перевищує попередні 23%. Нульову гіпотезу H₀ відхиляємо.
