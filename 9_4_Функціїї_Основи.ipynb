```python
def multipy_numbers(a, b):
    return a * b

pairs = [(10, 2), (7, 8), (1024, 999)]

for a, b in pairs:
    print(multipy_numbers(a, b))
```

```python
def int_division(a, b):
    if type(a) != int or type(b) != int:
        print("Не працюю з такими типами даних.")
    elif b == 0:
        return "Не можу виконати ділення на 0"
    else:
        return a / b

pairs = [(12, 6), (20, 0), (112, 10), (99, 0), (11, 1.1)]

for a, b in pairs:
    print("Результат роботи функції:", int_division(a, b))
```

None повертається у випадку, коли функція отримує аргумент не типу int. У цьому випадку функція лише виводить повідомлення за допомогою print(), але не має команди return. Якщо функція не виконує return, Python автоматично повертає None.
