```python
def is_even(number):
    return number % 2 == 0
```

```python
[is_even(digit) for digit in [101, 2002, 3043, 5015, 8222]]
```

```python
def sum_numbers(*args):
    return sum(args)
```

```python
number_sets = [
    (1, 2, 3, 4),
    (10, 20),
    (11, 12, 13),
    (),
    (-19, -164, 183)
]
```

```python
for numbers in number_sets:
    print(sum_numbers(*numbers))
```

```python
person1 = [80, 1.62]
person2 = [69, 1.53]
person3 = [80, 1.66]
person4 = [80, 1.79]
person5 = [72, 1.60]
```

```python
def calculate_bmi(person):
    return int(person[0] / person[1]**2)
```

```python
print("bmi {} = {}".format('person1', calculate_bmi(person1)))
print("bmi {} = {}".format('person2', calculate_bmi(person2)))
print("bmi {} = {}".format('person3', calculate_bmi(person3)))
print("bmi {} = {}".format('person4', calculate_bmi(person4)))
print("bmi {} = {}".format('person5', calculate_bmi(person5)))
```
